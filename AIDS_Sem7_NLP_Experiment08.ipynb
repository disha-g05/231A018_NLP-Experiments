```python
print("Disha Gawade 231A018")
documents =[
"Solar panels convert sunlight into clean electricity.",
"Renewable energy can be generated from the sun using solar cells.",
"Electric vehicles help reduce greenhouse gas emissions in cities.",
"Photovoltaic systems capture solar energy to power homes.",
"Baking a cake requires precise measurements of flour and sugar."
]
```

```python
print("Disha Gawade 231A018")
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd
```

```python
print("Disha Gawade 231A018")
for i, document in enumerate(documents):
  print(f"D{i+1}: {document}")
```

```python
print("Disha Gawade 231A018")
vectorize = TfidfVectorizer(stop_words='english')
tfidf_matrix = vectorize.fit_transform(documents)
print(tfidf_matrix.shape)
```

```python
print("Disha Gawade 231A018")
print(tfidf_matrix)
```

```python
print("Disha Gawade 231A018")
tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns = vectorize.get_feature_names_out(),
    index=["D1","D2","D3","D4","D5"]
)
tfidf_df.round(2)
```

```python
print("Disha Gawade 231A018")
words= vectorize.get_feature_names_out()
print(words)
```

```python
print("Disha Gawade 231A018")
print(len(words))
```

```python
print("Disha Gawade 231A018")
similarity_matrix = cosine_similarity(tfidf_matrix)
similarity_matrix
```

```python
print("Disha Gawade 231A018")
similarity_df = pd.DataFrame(
  similarity_matrix,
  index=["D1", "D2", "D3", "D4", "D5"],
  columns=["D1", "D2", "D3", "D4", "D5"]
)

similarity_df.round(3)
```

```python
print("Disha Gawade 231A018")
highest_similarity = -1
most_similar_pair = None

for i in range(len(documents)):
  for j in range(i + 1, len(documents)):

    similarity = similarity_matrix[i][j]

    if similarity > highest_similarity:
      highest_similarity = similarity
      most_similar_pair = (i, j)

print(
f"Most similar documents: "
f"D{most_similar_pair[0]+1} and D{most_similar_pair[1]+1}"
)
print(f"Similarity: {highest_similarity:.3f}")
```
