```python
print(" Disha Ganesh Gawade, 231A018")
lowercase = text.lower()
print(lowercase)
```

removing url, email, twitter handle

```python
print(" Disha Ganesh Gawade, 231A018")
import re
pattern=r'https?://www\.\S+|www\.\S+|\S+\@\S+|\@\S+'
text_url=re.sub(pattern, " ", lowercase)
print(text_url)
```

remove emojis

```python
print(" Disha Ganesh Gawade, 231A018")
!pip install emoji
```

```python
print(" Disha Ganesh Gawade, 231A018")
import emoji
text_emoji = emoji.replace_emoji(text_url, replace=" ")
print(text_emoji)
```

Remove Numbers

```python
print(" Disha Ganesh Gawade, 231A018")
pattern = r'\d+\S+'
text_number = re.sub(pattern, " ", text_emoji)
print(text_number)
```

To remove punctuation. First tokenize

```python
print(" Disha Ganesh Gawade, 231A018")
import nltk
nltk.download('punkt_tab')
```

```python
print(" Disha Ganesh Gawade, 231A018")
from nltk import word_tokenize
text_token = word_tokenize(text_number)
print(text_token)
```

```python
print(" Disha Ganesh Gawade, 231A018")
import string
punctuation = list(string.punctuation)
print(punctuation)
```

```python
print(" Disha Ganesh Gawade, 231A018")
print("With punctuation: ",text_token)
print()
text_punct=[]
for word in text_token:
  if word not in punctuation:
    text_punct.append(word)
print("Without punctuation: ",text_punct)
```

Rmove stopwords

```python
print(" Disha Ganesh Gawade, 231A018")
from nltk.corpus import stopwords
nltk.download('stopwords')
```

```python
print(" Disha Ganesh Gawade, 231A018")
print(stopwords.words('english'))
```

```python
print(" Disha Ganesh Gawade, 231A018")
text_clean = []
for word in text_punct:
  if word not in stopwords.words('english'):
    text_clean.append(word)
print(text_clean)
```

Stemming

```python
print(" Disha Ganesh Gawade, 231A018")
from nltk.stem import PorterStemmer
stem = PorterStemmer()
print("Befor stemming",text_clean)
text_stem = [stem.stem(i) for i in text_clean]
print("After stemming",text_stem)
```

```python
print(" Disha Ganesh Gawade, 231A018")
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')
```

Lemmatization

```python
print(" Disha Ganesh Gawade, 231A018")
lemma = WordNetLemmatizer()
print("Before lemmatization",text_clean)
print("After Stemming", text_stem)
text_lemma = [lemma.lemmatize(i) for i in text_clean]
print("After lemmatization",text_lemma)
```
