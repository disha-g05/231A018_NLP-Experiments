Import libraries

```python
print("Disha Gawade 231A018")
import nltk
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('tagsets_json')
nltk.download('maxent_ne_chunker_tab')
nltk.download('words')
from nltk import pos_tag, word_tokenize, ne_chunk
from nltk.chunk import RegexpParser
```

```python
print("Disha Gawade 231A018")
sentence = "The talented software engineer from Microsoft attended a global tech conference in Paris last month."
```

Tokenization and POS tagging

```python
print("Disha Gawade 231A018")
tokens = word_tokenize(sentence)
tagged = pos_tag(tokens)
tagged
```

```python
print("Disha Gawade 231A018")
nltk.help.upenn_tagset("DT")
nltk.help.upenn_tagset("VBD")
nltk.help.upenn_tagset("JJ")
nltk.help.upenn_tagset("NNP")
```

NER

```python
named_entities = ne_chunk(tagged)
print(named_entities)
```

```python
print("Disha Gawade 231A018")
for subtree in named_entities:
  if isinstance(subtree, nltk.Tree):
    entity = " ".join([word for word, tag in subtree.leaves()])
    label = subtree.label()
    print(f"Entity: {entity}, Label: {label}")
```

Part 2: Chunking.
defining a chunk grammar

```python
print("Disha Gawade 231A018")
grammer = "NP: {<DT>?<JJ>*<NN>}"
```

create a chunk parser and parse the sentence

```python
print("Disha Gawade 231A018")
chunk_parser = RegexpParser(grammer)
chunked = chunk_parser.parse(tagged)

print(chunked)
```

```python
print("Disha Gawade 231A018")
from nltk.tree import tree
chunked.pretty_print()
```
